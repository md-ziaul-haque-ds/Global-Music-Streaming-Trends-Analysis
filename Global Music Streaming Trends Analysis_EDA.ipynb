```python
# Univariate Analysis (Distribution of user ages)

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Heatmap : Distribution of Music Genres by Age Group

```python
# Distribution of Music Genres by Age Group

# Creating age groups
bins =[0,18,30,50,70,100]
labels = ["<18", "18-30", "31-50", "51-70", "70+" ]
df["Age Group"] = pd.cut(df["Age"], bins= bins, labels=labels, right=False)

# Grouping data by "Age Group" and "Top Genre"
genre_age_count = df.groupby(["Age Group", "Top Genre"], observed= False).size().unstack().fillna(0)

# creating the heat map
plt.figure(figsize=(10,5))
sns.heatmap(genre_age_count, annot =True, fmt="d", cmap="YlGnBu", cbar_kws= {"label" :"Number of listeners"})
plt.title("Heatmap : Distribution of Music Genres by Age Group")
plt.xlabel("Music Genre")
plt.ylabel("Age Group")
plt.show()


```

```python
# Relationship between Subscription type and Minutes streamed per day.
Subscription_type = df.groupby("Subscription Type")["Minutes Streamed Per Day"].count()
Subscription_type                           # Insight : This shows users with premium subscription is more
```

```python
# Relationship between Most Played Artist and Number of Songs Liked.
most_liked_artists = df.groupby("Most Played Artist") ["Number of Songs Liked"].sum().sort_values(ascending=False)
```

```python
# Displaying the top artists with the most liked songs
print(most_liked_artists.head(10))
```

```python
# plotting the top 10 most liked artists
plt.figure(figsize = (12,6))
most_liked_artists.head(10).plot(kind="bar", colormap="coolwarm", edgecolor="black")

# Labels and Titles
plt.xlabel("Artist")
plt.ylabel("Total Number of songs liked")
plt.title("The top 10 artists with most liked artists")
plt.xticks(rotation=45)
plt.show()
```

# Question: What type of subscription do each group prefer ?

```python
# Preferred Subscription Type with Age Group.
# Question: What type of subscription do each group prefer ?
subscription_preference = df.groupby(["Age Group", "Subscription Type"]).size().reset_index(name= "Count")
# sorting values for better visibility 
subscription_preference = subscription_preference.sort_values(["Age Group", "Count"], ascending=[True,False])
# set plot style
plt.figure(figsize=(10,6))
sns.barplot(x= "Age Group", y = "Count", hue="Subscription Type",data=subscription_preference,palette="viridis")

#labels and title
plt.xlabel("Age Group", fontsize=12)
plt.ylabel("Number of Users", fontsize=12)
plt.title("Preferred Subscription Type by Age Group", fontsize=14)
plt.legend(title= "Subscription Type")
plt.xticks(rotation=45)
plt.show()

```

```python
# Age versus genre 
age_vs_genre = df.groupby(["Top Genre"])["Age"].mean().sort_values()
age_vs_genre
plt.figure(figsize=(8,5))
sns.barplot(x=age_vs_genre.index, y=age_vs_genre.values, palette="coolwarm")
plt.xticks(rotation=45)
plt.xlabel("Music Genre")
plt.ylabel("Average Age of listerns")
plt.title("Impact of Age on Music Preferences")
plt.show()
```

```python
# Top 10 Most Streamed Artists (Bar Chart)
plt.figure(figsize=(12,6))
sns.countplot(y=df["Most Played Artist"], order=df["Most Played Artist"].value_counts().index[:10], palette="magma")
plt.xlabel("Number of Users")
plt.ylabel("Artists")
plt.title("Top 10 Most Streamed Artists")
plt.show()
```

# Most trending genre in the country

```python
# Most trending genre in the country
regional_genre_trends = df.groupby("Country")["Top Genre"].agg(lambda x:x.value_counts().idxmax())
regional_genre_df = pd.DataFrame(regional_genre_trends).reset_index()
# sort data by country for better visualization
regional_genre_df = regional_genre_df.sort_values(by="Country")
# Plot a bar chart for regional trends 
plt.figure(figsize = (12,6))
sns.barplot(y=regional_genre_df["Country"], x=regional_genre_df.index, hue=regional_genre_df["Top Genre"], dodge=False, palette="viridis")
plt.xlabel("Most Popular Genre")
plt.ylabel("Country")
plt.title("Regional trends in Music Streaming preferences")
plt.legend(title="Top Genre", bbox_to_anchor=(1.05,1), loc="upper left") # place legend outside
plt.xticks(rotation=45)
plt.show()
```

# Question : Indentify which artist has the highest repeat song rate in each country

```python
# Group by Country and Artist and calculate the average repeat song rate
country_artist_repeat_rate = df.groupby(["Country", "Most Played Artist"])["Repeat Song Rate (%)"].mean().reset_index()
# Find the top artist in each country based on the highest repeat song rate
top_artist_by_country= country_artist_repeat_rate.loc[country_artist_repeat_rate.groupby("Country")["Repeat Song Rate (%)"].idxmax()]
# sort the values for better readability 
top_artist_by_country = top_artist_by_country.sort_values(by="Repeat Song Rate (%)", ascending = False)

# set figure size
plt.figure(figsize=(12,7))


# Create a barplot
sns.barplot(data=top_artist_by_country, x= "Repeat Song Rate (%)", y="Country", hue="Most Played Artist", palette ="coolwarm")

# labels and title
plt.xlabel("Repeat Song Rate", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.title("Top Artist in each country by Repeat Song Rate", fontsize=12)
plt.legend(title= "Most Played Artist", bbox_to_anchor=(1.05,1), loc= "upper left")

# show the plot
plt.show()
```
